## Um gráfico rápido

Receita mensal, direto do catálogo — sem reprocessar nada, o dataset já foi
gerado pelo pipeline `relatorio`.

```python
import matplotlib.pyplot as plt

receita = catalog.load("receita_mensal")
receita.plot(x="mes", y="receita_total", kind="line", figsize=(10, 4), title="Receita mensal")
plt.tight_layout()
plt.show()
```
